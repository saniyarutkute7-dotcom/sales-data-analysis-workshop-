### Data Visualization

```python
# how do the total scale change in month

import matplotlib.pyplot as plt

month = (
    df.set_index("order_date")
    .resample("M")["total_amount"].sum()
)

plt.figure(figsize =(10,10))
plt.plot(month.index,month.values,marker="o")
plt.xlabel("Month")
plt.ylabel("Total Scales")
plt.title("Monthly Total Amount")
plt.show()

```

```python
cat = (
    df.groupby("category")["total_amount"].sum()
)

plt.figure(figsize=(10,5))
plt.bar(cat.index,cat.values)
plt.title("Sales by Category")
plt.xlabel('category')
plt.ylabel('Total Scales')
plt.xticks(rotation=90, ha="right")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

month = (
    df.set_index("order_date")
    .resample("M")["total_amount"].sum()
)

plt.figure(figsize =(10,5))
plt.plot(month.index,month.values,marker="o")
plt.xlabel("Month")
plt.ylabel("Total Scales")
plt.tight_layout()
plt.show()
```

```python
payment_count = df['payment_method'].value_counts()

plt.figure(figsize=(5,5))
plt.pie(payment_count.values,labels=payment_count.index)
plt.title('Order share by Payment Method')
plt.show()
```

```python
# Distribution of order quantity
plt.figure(figsize=(10,5))
plt.hist(df['quantity'],bins=5,edgecolor='black')
plt.title("Distribution of order quantity")
plt.xlabel("Quantity")
plt.ylabel("no of quantity")
plt.show()
```

```python
channel_sales = (
    df.groupby('sales_channel')['total_amount'].sum()
)

plt.figure(figsize=(10,5))
plt.barh(channel_sales.index,channel_sales.values)
plt.title("Sales by Channels")
plt.xlabel("Total Sales")
plt.ylabel("Sales channel")
plt.show()
```

```python
product_sales = (
    df.groupby('product', as_index=False)['total_amount']
      .sum()
      .sort_values('total_amount', ascending=False)
)

print(product_sales)
```

```python
#using Seaborn 
#Top product b y sales

import seaborn as sns

product_sales = (
    df.groupby('product', as_index=False)['total_amount']
    .sum()
    .sort_values('total_amount',ascending = False)
)

plt.figure(figsize=(10,5))
sns.barplot(data=product_sales,x='total_amount',y='product')
plt.title('Top product by sales')
plt.xlabel('Total sales')
plt.ylabel('product')
plt.show()
```

```python
sns.countplot(
    data = df,
    x = 'payment_method',
    order=df['payment_method'].value_counts().index
    )
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="category",y="unit_price")
plt.title("Unit Price Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Unit Price")
plt.tight_layout()
plt.show()
```
